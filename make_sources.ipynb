# Source systems: a customer CSV export and a product JSON catalog.

```python
import csv, json, random
from pathlib import Path
```

```python
rng = random.Random(42)
```

```python
src = Path("source")
src.mkdir(exist_ok=True)
```

```python
cities = ["Karachi", "Lahore", "Manila", " manila", "CEBU", "Dubai", "Singapore", "TORonto", ]
```

```python
with open(src / "customers.csv", "w", newline="") as f:
    w = csv.writer(f)
    w.writerow(["customer_id", "name", "email", "city", "signup_date"])
    for i in range(1, 2001):
        email = "" if i % 97 == 0 else f"user{i}@example.com"
        row = [i, f"Customer {i}", email, rng.choice(cities), f"2026-{rng.randint(1, 12):02d}-{rng.randint(1, 28):02d}"]
        w.writerow(row)
    
        if i % 158 == 0:
            w.writerow(row)
        
        categories = ["Laptops", "Phones", "Audio", "Accessories", "Gaming"]
        products = [{"sku": f"SKU-{i:03d}", "name": f"{categories[i % 5]} item {i}", "category": categories[i % 5], "price": round(rng.uniform(5, 900), 2)} for i in range(1, 41)]

        (src / "products.json").write_text(json.dumps(products, indent=2))
    
    print("wrote", src / "customers.csv", "and", src / "products.json")
        
```
