# Run sql file, against duckdb

```python
import sys, duckdb
from pathlib import Path
con = duckdb.connect("warehouse.duckdb")
```

```python
sql = "sql/staging.sql"
```

```python
sql_query = Path(sql).read_text(encoding="utf-8") if sql.endswith(".sql") else sql
```

```python
# Sample run
# sql = open(arg).read() if arg.endswith(".sql") else arg
result = con.sql(sql_query)
if result is not None:
    result.show()
```

```python
con.close()
```
